```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(r"C:\Users\arwaa\Downloads\ride_hailing_dataset.csv")

# Display the first five rows
df.head()
```

```python
# Dataset dimensions
print("Dataset shape:", df.shape)

# Column names and data types
print("\nDataset information:")
df.info()

# Missing values in each column
print("\nMissing values:")
print(df.isnull().sum())
```

```python
average_fare = df.groupby("service_type")["fare"].mean()

print("Average fare by service type:")
print(average_fare.round(2))
```

```python
trips_by_city = df["city"].value_counts()

print("Number of trips by city:")
print(trips_by_city)

print("\nCity with the most trips:", trips_by_city.idxmax())
print("Number of trips:", trips_by_city.max())
```

```python
average_distance = df["distance_km"].mean()

print("Average distance per trip:", round(average_distance, 2), "km")
```

```python
ratings_by_gender = df.groupby("gender")["trip_rating"].mean()

print("Average trip rating by gender:")
print(ratings_by_gender.round(2))
```

```python
average_fare = df.groupby("service_type")["fare"].mean().sort_values(ascending=False)

average_fare.plot(kind="bar", figsize=(8, 5))

plt.title("Average Fare by Service Type")
plt.xlabel("Service Type")
plt.ylabel("Average Fare")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
